```python
import numpy as np 
```

```python
np.random.seed(0) # To insure that the code generates the same exact sequence
                  # of random numbers at every run
X=np.random.randn(100,3)
true_W=np.array([[2],[-1],[0.5]])
y=X @ true_W + 0.1* np.random.randn(100,1)                                   
```

```python
W = np.random.randn(3,1)
b = np.zeros((1,)) # Neutre au depart 
```

```python
def relu(z):
    return np.maximum(0,z)
def relu_derivation(z):
    return (z>0).astype(float)

```

```python
def forward(X,W,b):
    Z=X@W+b
    A=relu(Z)
    return Z,A
```

```python
def loss_fct(y_pred,y_true):
    return np.mean((y_pred-y_true)**2)
Z,A=forward(X,W,b)
print("Initial Loss:",loss_fct(A,y))
```

Ajuster W et b -> Backpropagation : 

```python
def backpropagation(X,y,Z,A,W):
    m=X.shape[0]
    dA = 2*(A-y)/m # Gradient de la loss 
    dZ=dA*relu_derivation(Z) # Si Z>0 -> return 1 -> l'erreur dA continue en arrière 
                             # Si Z =< 0 -> return 0 -> Annuler le gradient -> Aucune correction a faire 
    dW=X.T@dZ # Associer chaque variable a son erreur 
    db=np.sum(dZ,axis=0) # Sommer l'erreur sur l'ensemble de l'axe des echantillons 0 car le biais impacte toutes les obs de maniere identique
    return dW,db # Vont permettre de mettre a jour le modele via la descente de gradient 
```

```python
def update (W,b,dW,db,lr):
    W=W-lr * dW 
    b= b - lr*db 
    return W , b     
```

```python
epochs= 1000 
lr=0.01
for i in range (epochs):
    Z,A= forward (X,W,b)
    loss=loss_fct(A,y)
    dW,db=backpropagation(X,y,Z,A,W)
    W,b=update(W,b,dW,db,lr)
if i%100==0: # Obtenir la loss apres chaque 100 pas 
    print (f"Epoch{i},Loss:{loss}")    
    
```

```python
print("True_Weights:",true_W)
print("\nWeights learned by the Model:",W)
print("\nLearned biais:" , b )
```
